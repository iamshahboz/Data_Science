# Pandas is a tool for data analysis and data manipulation

## Two main data structures in Pandas
1. series  - one dimentional arrays
2. data frames - two dimentional arrays


```python
import pandas as pd
import numpy as np
```

```python
# creating an array
data = np.array([[1,4], [2,5], [3,7]])
```

```python
# creating a dataframe
df = pd.DataFrame(data, index=['row1','row2','row3'], columns=['col1','col2'])
```

```python
df
```

# Creating a DataFrame from dictionary

```python
# lists

states = ['California','Taxes','Florida','New York']
population = [204456, 7123410, 1974887, 3551300]

dict_states = {'States': states, 'Population': population}
```

```python
df_population = pd.DataFrame(dict_states)
```

```python
df_population
```
